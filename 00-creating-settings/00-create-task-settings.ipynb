COME APPROCCIO INIZIALE IMPOSTIAMO MOLTI BENCHMARK IN MODO SIMILE

```python
# This function is used to rapidly edit the task-settings file and set a limit to all tasks
def set_limit_to_all(limit, filename):
    with open(filename, "r") as f:
        settings_dict = json.load(f)
    for task_id in settings_dict:
        settings_dict[task_id]["limit"] = limit
        add_setting(task_id, settings_dict[task_id], edit=True, settings_file=TASK_SETTINGS_FILE)

def set_limit(limit, task_id, filename=TASK_SETTINGS_FILE):
    with open(filename, "r") as f:
        task_settings = json.load(f)[task_id]
    task_settings["limit"] = limit
    add_setting(task_id, settings = task_settings, edit=True, settings_file=filename)

def add_setting(task, settings:dict, edit=False, settings_file=TASK_SETTINGS_FILE):
    with open(settings_file, "r") as f:
        settings_dict = json.load(f)
        if not edit and task in settings_dict.keys():
            raise Exception("add_setting not in edit mode! ")
        else:
            settings_dict[task] = settings
    with open(settings_file, "w") as f:
        json.dump(settings_dict, f, indent=4, ensure_ascii=False)
```

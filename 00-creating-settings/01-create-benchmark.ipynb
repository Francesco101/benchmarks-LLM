This notebook is used to construct benchmarks, just correlating models and tasks to their relative settings in the task-settings file.

```python
BENCHMARK_FILE = "../benchmarks/round-0-bench.json"
# The task-settings file the benchmark will have to refer to
TASK_SETTINGS_FILE = "../benchmarks/task-settings/round-0-task-settings.json"
```

```python
import json

MODELS = [
    "Qwen/Qwen3.6-35B-A3B-FP8",
    "Qwen/Qwen3.8-27B-FP8"
]


TASKS = [
    "winogrande",
    "truthfulqa_mc2",
    "minerva_math",
    "leaderboard_bbh",
    "gsm8k",
    "mmlu",
    "arc_challenge",
    "hellaswag",
]
TASK_SETTINGS = TASKS       # Because of how task-settings.json is currently structured, but it could be different
```

Construct benchmark instructions

```python
# This has sense for how task-settings file is currently constructed, not generally

bcm = {}
bcm["tasks_settings_file"] = TASK_SETTINGS_FILE
for model in MODELS:
    bcm[model] = {}
    bcm[model]["tasks"] = TASKS
    bcm[model]["task-settings"] = TASK_SETTINGS


with open(BENCHMARK_FILE, "w") as f:
    json.dump(bcm, f, indent=4, ensure_ascii=True)
```
